```python
'''
Calculate the following metrics department-wise:
1. Total number of male employees.
2. Total number of female employees.
3. Number of employees earning more than ₹70,000.
4. Number of employees earning less than ₹50,000.
5. Total salary paid to male employees.
6. Total salary paid to female employees.
'''


from pyspark.sql import functions as F

result = df.groupBy("department").agg(

    # Count male employees
    F.sum(
        F.when(F.col("gender") == "Male", 1).otherwise(0)
    ).alias("male_count"),

    # Count female employees
    F.sum(
        F.when(F.col("gender") == "Female", 1).otherwise(0)
    ).alias("female_count"),

    # Employees earning more than 70000
    F.sum(
        F.when(F.col("salary") > 70000, 1).otherwise(0)
    ).alias("salary_above_70k"),

    # Employees earning less than 50000
    F.sum(
        F.when(F.col("salary") < 50000, 1).otherwise(0)
    ).alias("salary_below_50k"),

    # Total salary paid to male employees
    F.sum(
        F.when(F.col("gender") == "Male", F.col("salary"))
         .otherwise(0)
    ).alias("male_total_salary"),

    # Total salary paid to female employees
    F.sum(
        F.when(F.col("gender") == "Female", F.col("salary"))
         .otherwise(0)
    ).alias("female_total_salary")
)

result.orderBy("department").show(truncate=False)
```

```python
'''
Problem 11 — Explode Nested Arrays of Structs and Calculate Product Revenue
ou receive order-level data. Each order contains multiple products, and every product has:
- Product name
- Quantity purchased
- Unit price
Your task is to:
1. Convert the nested product array into individual rows.
2. Calculate revenue for each product in each order.
3. Calculate total quantity sold per product.
4. Calculate total revenue per product.
5. Sort the output by total revenue in descending order.
'''
from pyspark.sql.types import (
    StructType, StructField,
    IntegerType, StringType,
    ArrayType, DoubleType
)

from pyspark.sql import functions as F

# Define the nested product structure
product_schema = StructType([
    StructField("product_name", StringType(), True),
    StructField("quantity", IntegerType(), True),
    StructField("unit_price", DoubleType(), True)
])

# Define the order-level schema
order_schema = StructType([
    StructField("order_id", IntegerType(), True),
    StructField("customer", StringType(), True),
    StructField("products", ArrayType(product_schema), True)
])

# Sample data
orders = [
    (1001, "Aman", [
        ("Laptop", 1, 60000.0),
        ("Mouse", 2, 800.0)
    ]),
    (1002, "Rahul", [
        ("Laptop", 1, 58000.0),
        ("Keyboard", 1, 1500.0)
    ]),
    (1003, "Priya", [
        ("Mouse", 1, 800.0),
        ("Monitor", 1, 12000.0)
    ])
]

orders_nested_df = spark.createDataFrame(
    orders,
    schema=order_schema
)

orders_nested_df.printSchema()
orders_nested_df.show(truncate=False)


exploded_df = orders_nested_df.withColumn(
    "product",
    F.explode(F.col("products"))
)

product_df = exploded_df.select(
    "order_id",
    "customer",
    F.col("product.product_name").alias("product_name"),
    F.col("product.quantity").alias("quantity"),
    F.col("product.unit_price").alias("unit_price")
)

product_df = product_df.withColumn(
    "revenue",
    F.col("quantity") * F.col("unit_price")
)

product_df.show()

# Product-wise aggregation
result_df = product_df.groupBy("product_name").agg(
    F.sum("quantity").alias("total_quantity"),
    F.sum("revenue").alias("total_revenue")
)

result_df = result_df.orderBy(
    F.desc("total_revenue")
)

result_df.show()
```

```python
# lag() and lead() in PySpark Window Functions

'''
Task 
1.Previous day's sales for each product.



2.Next day's sales for each product.



3.Difference between current sales and previous sales.



4.Percentage change compared with previous sales.

'''


from pyspark.sql import functions as F
from pyspark.sql.window import Window

data = [
    ("Laptop", "2026-01-01", 10000),
    ("Laptop", "2026-01-02", 15000),
    ("Laptop", "2026-01-03", 12000),
    ("Mobile", "2026-01-01", 5000),
    ("Mobile", "2026-01-02", 7000),
    ("Mobile", "2026-01-03", 9000),
]

product_sales_df = spark.createDataFrame(
    data, ["product", "sales_date", "sales"]
)

product_sales_df = product_sales_df.withColumn(
    "sales_date",
    F.to_date("sales_date")
)

product_sales_df.show()
```

```python
# define the window
window_spec = Window.partitionBy(
    "product"
).orderBy(
    "sales_date"
)

result_df = product_sales_df.withColumn(
    "previous_sales",
    F.lag("sales", 1).over(window_spec)
).withColumn(
    "next_sales",
    F.lead("sales", 1).over(window_spec)
)

result_df.orderBy(
    "product", "sales_date"
).show()
```

```python
# Step 4 — Calculate sales difference


# Sales Difference  = Current Sales - Previous Sales

result_df = result_df.withColumn(
    "sales_difference",
    F.col("sales") - F.col("previous_sales")
)

result_df.show()
```

```python
# Step 5 — Calculate percentage change

# Percentage Change = (Current Sales - Previous Sales)\Previous Sales X 100


result_df = result_df.withColumn(
    "percentage_change",
    F.when(
        F.col("previous_sales").isNull() |
        (F.col("previous_sales") == 0),
        F.lit(None).cast("double")
    ).otherwise(
        F.round(
            (
                (F.col("sales") - F.col("previous_sales"))
                / F.col("previous_sales")
            ) * 100,
            2
        )
    )
)

result_df.orderBy(
    "product", "sales_date"
).show(truncate=False)
```

```python
'''

Problem 16 — User Sessionization using lag() and Cumulative Sum
Advanced
Real-world Data Engineering

PROBLEM 16 OF 40
1. Business scenario
Imagine you're working for an e-commerce company like Amazon.

Users visit the website and perform different activities:
- View a product.
- Add an item to their cart.
- Make a purchase.
- Leave the website.
- Return after some time.
The business wants to identify individual user sessions.

Sessionization rule:

If a user has been inactive for more than 30 minutes, a new session starts.

Your task is to:
1. Identify the previous activity timestamp for each user.
2. Calculate the time difference between consecutive activities.
3. Identify when a new session starts.
4. Assign a session number to each activity.
5. Calculate the total number of sessions per user.

'''

# Step 1: Create a dataframe

from pyspark.sql import functions as F
from pyspark.sql.window import Window

events = [
    ("U1", "2026-01-01 10:00:00", "Page View"),
    ("U1", "2026-01-01 10:10:00", "Add to Cart"),
    ("U1", "2026-01-01 10:20:00", "Purchase"),
    ("U1", "2026-01-01 11:00:00", "Page View"),
    ("U1", "2026-01-01 11:15:00", "Purchase"),
    ("U2", "2026-01-01 09:00:00", "Page View"),
    ("U2", "2026-01-01 09:45:00", "Purchase"),
    ("U2", "2026-01-01 10:00:00", "Page View")
]

events_df = spark.createDataFrame(
    events,
    ["user_id", "event_timestamp", "event"]
)

events_df = events_df.withColumn(
    "event_timestamp",
    F.to_timestamp("event_timestamp")
)

events_df.show(truncate=False)
```

```python
#Step 2 — Identify previous activity
window_spec = Window.partitionBy(
    "user_id"
).orderBy(
    "event_timestamp"
)

# Now use lag():

events_df = events_df.withColumn(
    "previous_timestamp",
    F.lag("event_timestamp").over(window_spec)
)

# Why lag()?

# We need to compare the current event timestamp with the previous event timestamp for the same user.


# Step 3 — Calculate the inactivity gap

events_df = events_df.withColumn(
    "time_difference_minutes",
    (
        F.col("event_timestamp").cast("long") -
        F.col("previous_timestamp").cast("long")
    ) / 60
)

'''
# For example:
#- Previous activity: 10:20
#- Current activity: 11:00
#- Difference: 40 minutes.


# Step 4 — Identify new sessions


This is the important business rule:
- First activity for a user → New session.
- Gap greater than 30 minutes → New session.
- Gap of 30 minutes or less → Existing session.

'''
events_df = events_df.withColumn(
    "new_session",
    F.when(
        F.col("previous_timestamp").isNull() |
        (F.col("time_difference_minutes") > 30),
        1
    ).otherwise(0)
)

#Notice that exactly 30 minutes does not create a new session because our rule says more than 30 minutes.
```

```python
 # Step 5 — Assign session numbers
 # We use a cumulative sum of the new_session column.

session_window = (
    Window
    .partitionBy("user_id")
    .orderBy("event_timestamp")
    .rowsBetween(
        Window.unboundedPreceding,
        Window.currentRow
    )
)

events_df = events_df.withColumn(
    "session_number",
    F.sum("new_session").over(session_window)
)

events_df.orderBy(
    "user_id",
    "event_timestamp"
).show(truncate=False)
```

```python
# Final interview extension
session_summary = events_df.groupBy(
    "user_id"
).agg(
    F.max("session_number").alias("total_sessions")
)

session_summary.show()
```

```python
# Problem 17 — Identify Missing Dates and Fill Gaps in PySpark

'''

Your source data contains sales for some dates, but not all dates.

For example:
Date	Product	Sales
Jan 1	Laptop	1000
Jan 2	Laptop	2000
Jan 5	Laptop	5000

January 3 and January 4 are missing.

The business wants a continuous daily sales report, where missing dates have sales of zero.

Your task:
1. Identify missing dates.
2. Generate a continuous date range.
3. Create a row for every date.
4. Join the original sales data with the generated dates.
5. Replace missing sales with zero.

'''


# Step 1 — Create the DataFrame

from pyspark.sql import functions as F

sales_data = [
    ("2026-01-01", "Laptop", 1000),
    ("2026-01-02", "Laptop", 2000),
    ("2026-01-05", "Laptop", 5000),
    ("2026-01-01", "Mobile", 3000),
    ("2026-01-03", "Mobile", 4000),
    ("2026-01-05", "Mobile", 6000)
]

sales_df = spark.createDataFrame(
    sales_data,
    ["sales_date", "product", "sales"]
)

sales_df = sales_df.withColumn(
    "sales_date",
    F.to_date("sales_date")
)

sales_df.orderBy("product", "sales_date").show()
```

```python
# Step 2 — Generate a date range

# First, find the minimum and maximum date for each product.

date_range_df = sales_df.groupBy("product").agg(
    F.min("sales_date").alias("start_date"),
    F.max("sales_date").alias("end_date")
)

date_range_df.show()
```

```python
# Now generate every date between the start and end dates:

calendar_df = date_range_df.withColumn(
    "sales_date",
    F.explode(
        F.sequence(
            F.col("start_date"),
            F.col("end_date"),
            F.expr("INTERVAL 1 DAY")
        )
    )
)

calendar_df = calendar_df.select(
    "product",
    "sales_date"
)

calendar_df.orderBy("product", "sales_date").show()
```

```python
'''

Understand these three functions
Function	Purpose
sequence()	Generates a sequence of dates
explode()	Converts each generated date into a separate row
groupBy()	Finds the date range independently for each product

For Laptop, sequence() generates January 1 through January 5.

For Mobile, it also generates January 1 through January 5.

'''

# Step 3 — Left join with original sales
# Now we have a complete calendar, but it doesn't contain sales values.

# We need to join it with the original DataFrame.

result_df = calendar_df.join(
    sales_df,
    on=["product", "sales_date"],
    how="left"
)


# Why left join?

# Because we want to preserve every date from our generated calendar, even when no corresponding sales record exists.

# Step 4 — Replace NULL with zero

result_df = result_df.withColumn(
    "sales",
    F.coalesce(F.col("sales"), F.lit(0))
)


# coalesce() returns the first non-NULL value.

# For example:
# - Sales = 1000 → returns 1000.
# - Sales = NULL → returns 0.

#    final Solution
result_df.select(
    "sales_date",
    "product",
    "sales"
).orderBy(
    "product",
    "sales_date"
).show()
```

```python
'''
Interview insights
This problem tests several important Data Engineering concepts:
- Date spine generation: Creating a continuous calendar for reporting.
- sequence() and explode(): Generating rows dynamically instead of manually creating dates.
- Left join: Preserving records from the calendar.
- coalesce(): Handling missing values.
- Partitioning by product: Generating separate date ranges for each product.
One important production consideration: if your source contains multiple sales transactions per product per day, aggregate them first. Otherwise, the join may produce duplicate calendar dates.

Also, this solution fills dates only between each product's own minimum and maximum dates. A business requirement to fill a common reporting period would need a different calendar boundary.

'''
```

```python
# Problem 18 — Data Deduplication Using Multiple Conditions

'''

1. Business scenario
Imagine you're working on a banking data pipeline.

Your source system sends customer transaction records, but due to retries and ingestion issues, duplicate transactions appear.

Your task is to:
1. Identify duplicate transaction IDs.
2. Retain only the latest transaction record.
3. If timestamps are equal, retain the record with the highest ingestion ID.
4. Produce a clean DataFrame.
This is a very common Data Engineering interview scenario.
2. Sample input
Transaction ID	Customer	Amount	Updated at	Ingestion ID
T101	Rahul	1000	10:00	1
T102	Priya	2000	10:05	2
T101	Rahul	1500	10:10	3
T103	Aman	3000	10:15	4
T102	Priya	2500	10:20	5
T101	Rahul	1800	10:10	6

Notice:
- T101 occurs three times.
- T102 occurs twice.
- T103 occurs once.
- T101 has two records with the same timestamp, so we need a tie-breaker.

'''

# Create a DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

transactions = [
    ("T101", "Rahul", 1000, "2026-01-01 10:00:00", 1),
    ("T102", "Priya", 2000, "2026-01-01 10:05:00", 2),
    ("T101", "Rahul", 1500, "2026-01-01 10:10:00", 3),
    ("T103", "Aman", 3000, "2026-01-01 10:15:00", 4),
    ("T102", "Priya", 2500, "2026-01-01 10:20:00", 5),
    ("T101", "Rahul", 1800, "2026-01-01 10:10:00", 6)
]

transactions_df = spark.createDataFrame(
    transactions,
    [
        "transaction_id",
        "customer",
        "amount",
        "updated_at",
        "ingestion_id"
    ]
)

transactions_df = transactions_df.withColumn(
    "updated_at",
    F.to_timestamp("updated_at")
)

transactions_df.show(truncate=False)
```

```python
# Step 2 — Understand the ranking logic

# We need the latest record for every transaction ID.

# Our ordering rule:
# 1. Latest updated_at first.
# 2. Highest ingestion_id first when timestamps are equal.


window_spec = Window.partitionBy(
    "transaction_id"
).orderBy(
    F.col("updated_at").desc(),
    F.col("ingestion_id").desc()
)

# Why use two columns in orderBy()?

# Because timestamps can be identical. The ingestion ID makes our ordering deterministic.

#Step 3 — Apply row_number()


ranked_df = transactions_df.withColumn(
    "row_num",
    F.row_number().over(window_spec)
)

ranked_df.orderBy(
    "transaction_id",
    "row_num"
).show(truncate=False)
```

```python
# Step 4 — Remove duplicates
deduplicated_df = ranked_df.filter(
    F.col("row_num") == 1
).drop("row_num")

deduplicated_df.show(truncate=False)
```

```python
'''

Important interview comparison
Method	Behaviour
dropDuplicates(["transaction_id"])	Keeps an arbitrary record per transaction ID; does not guarantee latest
row_number()	Allows controlled selection of exactly one record
dense_rank()	Retains all records tied at the highest rank
rank()	Also retains ties, with gaps in subsequent ranking values

Interview rule: If the interviewer asks you to retain the latest record based on a timestamp, use a window function with row_number() and a deterministic tie-breaker

'''
# Advanced challenge — Data quality checks

print("Original count:", transactions_df.count())

print("Final count:", deduplicated_df.count())

print(
    "Unique transaction IDs:",
    deduplicated_df.select("transaction_id").distinct().count()
)
```

```python
# Problem 20 — Find Consecutive Login Days Using PySpark

'''

2. Business scenario
Imagine you're working for a streaming platform such as Netflix.

The business wants to identify users who logged in for at least 3 consecutive calendar days.

You have a login activity dataset.

Your task is to:
1. Remove duplicate logins on the same day.
2. Identify consecutive login streaks for each user.
3. Calculate the length of each streak.
4. Return users with streaks of at least 3 days.
Input data
User	Login date
U1	Jan 1
U1	Jan 2
U1	Jan 3
U1	Jan 5
U1	Jan 6
U2	Jan 1
U2	Jan 3
U2	Jan 4
U2	Jan 5
U3	Jan 2
U3	Jan 3
'''
# 3. Step 1 — Create the DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

login_data = [
    ("U1", "2026-01-01"),
    ("U1", "2026-01-02"),
    ("U1", "2026-01-03"),
    ("U1", "2026-01-03"),  # Duplicate login
    ("U1", "2026-01-05"),
    ("U1", "2026-01-06"),
    ("U2", "2026-01-01"),
    ("U2", "2026-01-03"),
    ("U2", "2026-01-04"),
    ("U2", "2026-01-05"),
    ("U3", "2026-01-02"),
    ("U3", "2026-01-03")
]

login_df = spark.createDataFrame(
    login_data,
    ["user_id", "login_date"]
)

login_df = login_df.withColumn(
    "login_date",
    F.to_date("login_date")
)

login_df.show()
```

```python
# Step 2 — Remove duplicate login dates
# A user might log in multiple times on the same day. We only want one record per user per calendar day.

unique_logins_df = login_df.dropDuplicates(
    ["user_id", "login_date"]
)

# Interview insight: Without deduplication, the same day could receive multiple row numbers and break our streak logic.
```

```python
# Step 3 — Assign row numbers

window_spec = Window.partitionBy(
    "user_id"
).orderBy(
    "login_date"
)

ranked_df = unique_logins_df.withColumn(
    "row_num",
    F.row_number().over(window_spec)
)

ranked_df.orderBy(
    "user_id", "login_date"
).show()
```

```python
# Step 4 — Create a streak identifier

# For consecutive dates, subtracting the row number from the date produces the same reference date.

streak_df = ranked_df.withColumn(
    "streak_group",
    F.expr("date_sub(login_date, row_num)")
)

streak_df.orderBy(
    "user_id", "login_date"
).show()
```

```python
'''

Understand the logic
Login date	Row number	Date minus row number
Jan 1	1	Dec 31
Jan 2	2	Dec 31
Jan 3	3	Dec 31
Jan 5	4	Jan 1
Jan 6	5	Jan 1

Notice:
- January 1–3 have the same streak_group.
- January 5–6 have another streak_group.
This is the core Gaps and Islands technique.

'''

# Step 5 — Calculate streak lengths

streak_summary_df = streak_df.groupBy(
    "user_id",
    "streak_group"
).agg(
    F.min("login_date").alias("streak_start"),
    F.max("login_date").alias("streak_end"),
    F.count("*").alias("consecutive_days")
)

streak_summary_df.orderBy(
    "user_id", "streak_start"
).show()
```

```python
# Step 6 — Find streaks of at least 3 days

result_df = streak_summary_df.filter(
    F.col("consecutive_days") >= 3
).select(
    "user_id",
    "streak_start",
    "streak_end",
    "consecutive_days"
).orderBy(
    "user_id"
)

result_df.show(truncate=False)
```

```python
# Problem 21 — Find the Longest Consecutive Login Streak for Each User

'''

Interview scenario
You are working for a company that tracks user activity on its platform.

You receive login records from multiple users. The business wants to identify the longest consecutive login streak for every user.

Unlike Problem 20, where we found streaks of at least three days, now we need to:
1. Remove duplicate login dates.
2. Identify all consecutive login streaks.
3. Calculate the duration of each streak.
4. Find the longest streak for each user.
5. If two streaks have the same duration, return the most recent streak.
This is a multi-step interview problem.
2. Input dataset
User	Login date
U1	Jan 1
U1	Jan 2
U1	Jan 3
U1	Jan 5
U1	Jan 6
U1	Jan 7
U1	Jan 8
U2	Jan 1
U2	Jan 2
U2	Jan 4
U2	Jan 5
U2	Jan 6
U3	Jan 1
U3	Jan 3
U3	Jan 5

All dates are in January 2026.
3. Expected output
User	Longest streak start	Longest streak end	Days
U1	Jan 5	Jan 8	4
U2	Jan 4	Jan 6	3
U3	Jan 5	Jan 5	1

Notice that U3 has three separate one-day streaks. Since all have equal duration, we select the most recent one.
'''
# Step 1 — Create the DataFrame

from pyspark.sql import functions as F
from pyspark.sql.window import Window

login_data = [
    ("U1", "2026-01-01"),
    ("U1", "2026-01-02"),
    ("U1", "2026-01-03"),
    ("U1", "2026-01-05"),
    ("U1", "2026-01-06"),
    ("U1", "2026-01-07"),
    ("U1", "2026-01-08"),
    ("U2", "2026-01-01"),
    ("U2", "2026-01-02"),
    ("U2", "2026-01-04"),
    ("U2", "2026-01-05"),
    ("U2", "2026-01-06"),
    ("U3", "2026-01-01"),
    ("U3", "2026-01-03"),
    ("U3", "2026-01-05"),
    ("U1", "2026-01-03")  # Duplicate
]

login_df = spark.createDataFrame(
    login_data,
    ["user_id", "login_date"]
)

login_df = login_df.withColumn(
    "login_date",
    F.to_date("login_date")
)
```

```python
# Step 2 — Remove duplicate login dates
unique_logins_df = login_df.dropDuplicates(
    ["user_id", "login_date"]
)

# Why?

# A user may log in several times on the same date. We count that as one calendar-day login.

# Step 3 — Create the streak identifier

window_spec = Window.partitionBy(
    "user_id"
).orderBy(
    "login_date"
)

ranked_df = unique_logins_df.withColumn(
    "row_num",
    F.row_number().over(window_spec)
)

streak_df = ranked_df.withColumn(
    "streak_group",
    F.expr("date_sub(login_date, row_num)")
)
```

```python
'''
Why does this work?

For consecutive dates, subtracting their sequential row numbers produces the same date.

For example:
Date	Row number	Streak group
Jan 1	1	Dec 31
Jan 2	2	Dec 31
Jan 3	3	Dec 31
Jan 5	4	Jan 1
Jan 6	5	Jan 1

Same group means same consecutive streak.
'''

# Step 4 — Calculate each streak's duration

streak_summary_df = streak_df.groupBy(
    "user_id",
    "streak_group"
).agg(
    F.min("login_date").alias("streak_start"),
    F.max("login_date").alias("streak_end"),
    F.count("*").alias("streak_days")
)

# At this stage, we have all streaks, not just the longest ones.
# Step 5 — Find the longest streak per user

# Now comes the new concept: ranking aggregated results using another window.

ranking_window = Window.partitionBy(
    "user_id"
).orderBy(
    F.col("streak_days").desc(),
    F.col("streak_end").desc()
)

ranked_streaks_df = streak_summary_df.withColumn(
    "rank",
    F.row_number().over(ranking_window)
)



# Understand the ordering:
# 1. streak_days DESC: Longest streak first.
# 2. streak_end DESC: Most recent streak first when durations tie.
# 3. row_number(): Assigns exactly one first-place record per user.

# Step 6 — Get the final result

result_df = ranked_streaks_df.filter(
    F.col("rank") == 1
).select(
    "user_id",
    "streak_start",
    "streak_end",
    "streak_days"
)

result_df.orderBy("user_id").show()
```

```python
'''

Problem 22 — Detect INSERT, UPDATE and UNCHANGED Records
Advanced
ETL / Data Warehousing

ADVANCED TRACK · PROBLEM 22 OF 30
1. Business scenario
You work on a customer data pipeline.

You receive a new source dataset every day, and you already have customer records in your target table.

You need to compare both datasets and identify:
- INSERT: Customer exists in source but not in target.
- UPDATE: Customer exists in both, but one or more attributes changed.
- UNCHANGED: Customer exists in both and no attributes changed.
- DELETE: Customer exists in target but not in source.
This is a common ETL change-detection requirement.
2. Input datasets
Existing target records
ID	Name	City	Salary
101	Rahul	Mumbai	50000
102	Priya	Pune	60000
103	Aman	Delhi	70000
104	Sneha	Mumbai	55000

Incoming source records
ID	Name	City	Salary
101	Rahul	Mumbai	50000
102	Priya	Mumbai	65000
103	Aman	Delhi	70000
105	Rohan	Bangalore	45000

Expected classification:
ID	Status
101	UNCHANGED
102	UPDATE
103	UNCHANGED
104	DELETE
105	INSERT

'''

# Create the DataFrames

from pyspark.sql import functions as F

target_data = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", "Pune", 60000),
    (103, "Aman", "Delhi", 70000),
    (104, "Sneha", "Mumbai", 55000)
]

source_data = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", "Mumbai", 65000),
    (103, "Aman", "Delhi", 70000),
    (105, "Rohan", "Bangalore", 45000)
]

columns = ["customer_id", "name", "city", "salary"]

target_df = spark.createDataFrame(target_data, columns)
source_df = spark.createDataFrame(source_data, columns)

target_df.show()
source_df.show()
```

```python
'''
Step 2 — Perform FULL OUTER JOIN
Why full outer join?

Because we need to retain:
- Records present in both datasets.
- Records present only in source.
- Records present only in target.
'''

joined_df = source_df.alias("s").join(
    target_df.alias("t"),
    F.col("s.customer_id") == F.col("t.customer_id"),
    "full_outer"
)
```

```python
# Step 3 — Classify each record
result_df = joined_df.select(
    F.coalesce(
        F.col("s.customer_id"),
        F.col("t.customer_id")
    ).alias("customer_id"),

    F.coalesce(
        F.col("s.name"),
        F.col("t.name")
    ).alias("name"),

    F.when(
        F.col("t.customer_id").isNull(),
        "INSERT"
    ).when(
        F.col("s.customer_id").isNull(),
        "DELETE"
    ).when(
        (F.col("s.name") != F.col("t.name")) |
        (F.col("s.city") != F.col("t.city")) |
        (F.col("s.salary") != F.col("t.salary")),
        "UPDATE"
    ).otherwise("UNCHANGED").alias("record_status")
)
```

```python
'''
Important: Handle NULL values correctly
The comparison above needs improvement for production data.

For example:
- Source city = NULL
- Target city = Mumbai
A normal comparison involving NULL may evaluate to NULL rather than TRUE or FALSE.

Use Spark's null-safe equality operator eqNullSafe():
'''

changed = (
    ~F.col("s.name").eqNullSafe(F.col("t.name")) |
    ~F.col("s.city").eqNullSafe(F.col("t.city")) |
    ~F.col("s.salary").eqNullSafe(F.col("t.salary"))
)

# Then use this condition in your classification:

result_df = joined_df.select(
    F.coalesce(
        F.col("s.customer_id"),
        F.col("t.customer_id")
    ).alias("customer_id"),

    F.when(
        F.col("t.customer_id").isNull(),
        "INSERT"
    ).when(
        F.col("s.customer_id").isNull(),
        "DELETE"
    ).when(
        changed,
        "UPDATE"
    ).otherwise("UNCHANGED").alias("record_status")
)

result_df.orderBy("customer_id").show()


'''

Interview insights
Concept	Why it matters
Full outer join	Finds records on both sides, including unmatched records
coalesce()	Selects the available customer ID
when()	Implements ETL classification rules
eqNullSafe()	Compares values correctly when NULLs exist
Aliases	Avoids ambiguity between source and target columns

Real-world application: This pattern is used in incremental data loads, data reconciliation, change detection and warehouse pipelines.

One important production rule: source and target should have unique customer IDs before this join. Duplicate keys can multiply rows and produce incorrect classifications

'''
```

```python
# Problem 23 — Incremental Data Processing Using Hashing in PySpark
'''
This is an important Data Engineer interview problem. It builds directly on Problem 22, where we classified records as INSERT, UPDATE, DELETE and UNCHANGED.

Today, we'll use hashing to detect changed records efficiently.
1. Business scenario
Imagine you work for a banking company.

Your source system sends 10 million customer records every day.

You don't want to compare every individual column between the source and target tables manually.

Instead, you generate a hash for each customer's attributes.
- Same hash → Record unchanged.
- Different hash → Record potentially updated.
- Source-only customer → INSERT.
- Target-only customer → DELETE.


. Input datasets
Existing target records
ID	Name	City	Salary
101	Rahul	Mumbai	50000
102	Priya	Pune	60000
103	Aman	Delhi	70000
104	Sneha	Mumbai	55000

Incoming source records
ID	Name	City	Salary
101	Rahul	Mumbai	50000
102	Priya	Mumbai	65000
103	Aman	Delhi	70000
105	Rohan	Bangalore	45000

Our expected classification:
ID	Status
101	UNCHANGED
102	UPDATE
103	UNCHANGED
104	DELETE
105	INSERT

'''

# Create the DataFrames

from pyspark.sql import functions as F

target_data = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", "Pune", 60000),
    (103, "Aman", "Delhi", 70000),
    (104, "Sneha", "Mumbai", 55000)
]

source_data = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", "Mumbai", 65000),
    (103, "Aman", "Delhi", 70000),
    (105, "Rohan", "Bangalore", 45000)
]

columns = ["customer_id", "name", "city", "salary"]

target_df = spark.createDataFrame(target_data, columns)
source_df = spark.createDataFrame(source_data, columns)
```

```python
# Step 2 — Generate SHA-256 hashes
# PySpark provides a built-in function called sha2().

target_hashed_df = target_df.withColumn(
    "record_hash",
    F.sha2(
        F.concat_ws(
            "||",
            F.coalesce(F.col("name"), F.lit("<NULL>")),
            F.coalesce(F.col("city"), F.lit("<NULL>")),
            F.col("salary").cast("string")
        ),
        256
    )
)

# Now generate the hash for the source:

source_hashed_df = source_df.withColumn(
    "record_hash",
    F.sha2(
        F.concat_ws(
            "||",
            F.coalesce(F.col("name"), F.lit("<NULL>")),
            F.coalesce(F.col("city"), F.lit("<NULL>")),
            F.col("salary").cast("string")
        ),
        256
    )
)

'''
Understand the code
Function	Purpose
concat_ws()	Combines multiple column values into one string
coalesce()	Handles NULL values
cast("string")	Converts numeric values into strings
sha2()	Generates SHA-256 hash

For example:
Rahul||Mumbai||50000

Gets converted into a SHA-256 hash.

If the salary changes to 60000, the hash changes.

Important production detail: The simple delimiter-based serialization above is for this exercise. For robust production hashing, use unambiguous serialization (such as a canonical JSON representation or length-prefixed fields), consistent NULL handling and explicit data types. Otherwise, delimiters or special values can create ambiguous inputs.

'''

# Step 3 — Compare source and target hashes
# Now perform a full outer join.

joined_df = source_hashed_df.alias("s").join(
    target_hashed_df.alias("t"),
    on="customer_id",
    how="full_outer"
)

# Step 4 — Identify changes
# This is the main interview solution:

result_df = joined_df.select(
    F.coalesce(
        F.col("s.customer_id"),
        F.col("t.customer_id")
    ).alias("customer_id"),

    F.when(
        F.col("t.customer_id").isNull(),
        "INSERT"
    ).when(
        F.col("s.customer_id").isNull(),
        "DELETE"
    ).when(
        F.col("s.record_hash") != F.col("t.record_hash"),
        "UPDATE"
    ).otherwise(
        "UNCHANGED"
    ).alias("record_status")
)

result_df.orderBy("customer_id").show()

'''

Important interview concepts
Why hashing instead of comparing individual columns?

Suppose your customer table contains 50 attributes.

Without hashing, you may need to write 50 comparison conditions.

With hashing, you can compare one hash column.

Does hashing eliminate the join?

No. You still need to identify matching customer IDs. Hashing simplifies change detection after matching records.

Is SHA-256 collision-free?

No hash function is mathematically collision-free. SHA-256 has an extremely low collision probability and is suitable for this kind of change detection.

Does hashing always improve performance?

Not necessarily. Hash generation itself consumes CPU. Benchmark it against direct comparisons, especially for narrow datasets.
9. Production-level improvements
For an actual incremental pipeline:
- Generate hashes only for relevant business columns.
- Exclude technical fields such as ingestion timestamp and processing ID.
- Normalize strings, timestamps and NULLs consistently.
- Handle duplicate primary keys before joining.
- Use Delta Lake MERGE INTO to apply inserts and updates.
- Handle deletes according to the source's deletion semantics.

'''
```

```python
'''
Problem 24 — Delta Lake MERGE (Upsert)
Advanced
Interview relevant

Business scenario: You maintain a customer table in a data lake. Every day, a new batch arrives.

Your requirement:
- If customer_id already exists, update that customer's details.
- If customer_id is new, insert the record.
- Existing records not present in the incoming batch should remain unchanged.
This is called an upsert — update + insert.
'''
# Create the target Delta table

from pyspark.sql import functions as F
from delta.tables import DeltaTable

target_data = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", "Pune", 60000),
    (103, "Aman", "Delhi", 70000)
]

columns = ["customer_id", "name", "city", "salary"]

target_df_24 = spark.createDataFrame(target_data, columns)

target_df_24.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("customer_target_24")

# Create the incoming source batch

source_data = [
    (102, "Priya", "Mumbai", 65000),
    (103, "Aman", "Delhi", 70000),
    (104, "Sneha", "Pune", 55000)
]

source_df_24 = spark.createDataFrame(source_data, columns)

# Notice:
# - Customer 102: city and salary changed.
# - Customer 103: no changes.
# - Customer 104: new customer.
# - Customer 101: absent from source, but we will retain it.

# Perform the MERGE operation

delta_table_24 = DeltaTable.forName(
    spark,
    "customer_target_24"
)

delta_table_24.alias("target").merge(
    source_df_24.alias("source"),
    "target.customer_id = source.customer_id"
).whenMatchedUpdateAll() \
 .whenNotMatchedInsertAll() \
 .execute()

# Read the final Delta table

final_df_24 = spark.table("customer_target_24")

final_df_24.orderBy("customer_id").show()
```

```python
'''
Understand the MERGE logic
Clause	What it does
whenMatchedUpdateAll()	Updates all target columns when the key matches
whenNotMatchedInsertAll()	Inserts all columns when the key doesn't exist
execute()	Executes the MERGE transaction

Important interview concepts:
- Delta Lake MERGE supports atomic updates and inserts.
- You don't need to overwrite the entire target table for every incoming batch.
- The source should ideally contain only one record per merge key. Duplicate source keys can cause a MERGE failure because multiple source rows may match the same target row.
- This example does not delete customer 101 because we haven't specified any deletion condition.

'''
```

```python
# Problem 25 — Implement CDC (Change Data Capture) using Delta Lake MERGE
'''

Concepts: Delta Lake MERGE, INSERT, UPDATE, DELETE, CDC operations, conditional merge.
1. Business scenario
Imagine you're working on a banking data pipeline. Your source system sends customer changes every day.

Each incoming record contains an operation:
- I — Insert a new customer.
- U — Update an existing customer.
- D — Delete an existing customer.
Your task is to apply these changes to the target Delta table.

'''

# Create the target table
from pyspark.sql import functions as F
from delta.tables import DeltaTable

target_data_25 = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", "Pune", 60000),
    (103, "Aman", "Delhi", 70000),
    (104, "Sneha", "Mumbai", 55000)
]

columns_25 = ["customer_id", "name", "city", "salary"]

target_df_25 = spark.createDataFrame(
    target_data_25,
    columns_25
)

target_df_25.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("customer_cdc_target_25")

# Create the incoming CDC records

source_data_25 = [
    (102, "Priya", "Mumbai", 65000, "U"),
    (103, "Aman", "Delhi", 70000, "U"),
    (104, "Sneha", "Mumbai", 55000, "D"),
    (105, "Rohan", "Bangalore", 45000, "I")
]

source_columns_25 = [
    "customer_id",
    "name",
    "city",
    "salary",
    "operation"
]

source_df_25 = spark.createDataFrame(
    source_data_25,
    source_columns_25
)

source_df_25.show()

# Apply CDC using MERGE

delta_table_25 = DeltaTable.forName(
    spark,
    "customer_cdc_target_25"
)

delta_table_25.alias("target").merge(
    source_df_25.alias("source"),
    "target.customer_id = source.customer_id"
) \
.whenMatchedDelete(
    condition="source.operation = 'D'"
) \
.whenMatchedUpdate(
    condition="source.operation = 'U'",
    set={
        "name": "source.name",
        "city": "source.city",
        "salary": "source.salary"
    }
) \
.whenNotMatchedInsert(
    condition="source.operation = 'I'",
    values={
        "customer_id": "source.customer_id",
        "name": "source.name",
        "city": "source.city",
        "salary": "source.salary"
    }
) \
.execute()

final_df_25 = spark.table("customer_cdc_target_25")

final_df_25.orderBy("customer_id").show()


'''

Customer 104 was deleted, 102 was updated, 103 remained unchanged, and 105 was inserted.
6. Understand the MERGE clauses
Clause	Purpose
whenMatchedDelete()	Deletes matching target records
whenMatchedUpdate()	Updates matching target records
whenNotMatchedInsert()	Inserts new records
execute()	Executes the transaction

Interview insight: This is a common pattern in CDC-based ETL pipelines. In production, you must also handle duplicate source events, event ordering, retries, and late-arriving records. A production CDC pipeline often uses a source sequence number or event timestamp to prevent an older event from overwriting a newer state.

'''
```

```python
'''

Problem 26 — Handle Data Skew using Salting in PySpark
Advanced
Data Engineer interview topic

Concepts: Data skew, shuffle, partition imbalance, salting, joins, Spark optimization.

This is an important topic for Data Engineer interviews because a Spark job can be slow even when most partitions finish quickly.
1. Business scenario
Imagine you're processing an e-commerce orders dataset and joining it with customer details.

Your orders table has:
- Customer C001: 90,000 orders.
- Customer C002: 500 orders.
- Customer C003: 300 orders.
- Other customers: relatively few orders.
When Spark performs a join on customer_id, the large number of records belonging to C001 can create a heavily loaded partition.

This is called data skew.

'''

# Create sample data

from pyspark.sql import functions as F

# Orders dataset
orders_data_26 = [
    (i, "C001", 100 + i)
    for i in range(1, 16)
] + [
    (16, "C002", 200),
    (17, "C002", 300),
    (18, "C003", 400),
    (19, "C004", 500),
    (20, "C005", 600)
]

orders_df_26 = spark.createDataFrame(
    orders_data_26,
    ["order_id", "customer_id", "amount"]
)

# Customer dimension dataset
customers_data_26 = [
    ("C001", "Rahul"),
    ("C002", "Priya"),
    ("C003", "Aman"),
    ("C004", "Sneha"),
    ("C005", "Rohan")
]

customers_df_26 = spark.createDataFrame(
    customers_data_26,
    ["customer_id", "customer_name"]
)

# Understand the skew
orders_df_26.groupBy("customer_id") \
    .count() \
    .orderBy(F.desc("count")) \
    .show()
```

```python
# In a real dataset, the difference could be millions of records versus a few hundred.

# Perform a normal join

normal_join_26 = orders_df_26.join(
    customers_df_26,
    on="customer_id",
    how="inner"
)

normal_join_26.show()

'''

This is logically correct, but a heavily skewed key can make some shuffle partitions much larger than others.
5. Solve data skew using salting
What is salting?

Salting distributes records belonging to a hot key across multiple artificial keys.

For example, instead of sending every C001 record to the same join partition, we distribute it across five salt values:
- C001 + 0
- C001 + 1
- C001 + 2
- C001 + 3
- C001 + 4

'''
```

```python
# Add salt to the orders DataFrame
orders_salted_26 = orders_df_26.withColumn(
    "salt",
    F.when(
        F.col("customer_id") == "C001",
        F.pmod(F.xxhash64("order_id"), F.lit(5))
    ).otherwise(F.lit(0))
)

# Why use xxhash64?
# - It generates a deterministic hash.
# - We use pmod to produce values from 0 to 4.
# - Different order IDs are distributed across salt buckets.

# Replicate the hot key in the customer DataFrame
# The dimension table must contain all five salt values for C001, otherwise some orders won't find a matching customer.

customers_salted_26 = customers_df_26.withColumn(
    "salt_values",
    F.when(
        F.col("customer_id") == "C001",
        F.array(
            F.lit(0),
            F.lit(1),
            F.lit(2),
            F.lit(3),
            F.lit(4)
        )
    ).otherwise(
        F.array(F.lit(0))
    )
).withColumn(
    "salt",
    F.explode("salt_values")
).drop("salt_values")

# Join using customer_id and salt

salted_join_26 = orders_salted_26.join(
    customers_salted_26,
    on=["customer_id", "salt"],
    how="inner"
)

# Display the output
salted_join_26.select(
    "order_id",
    "customer_id",
    "customer_name",
    "amount"
).orderBy("order_id").show()
```

```python
'''
Compare normal join and salted join
Feature	Normal join	Salted join
Join keys	customer_id	customer_id + salt
Hot key distribution	Can be uneven	Distributed across buckets
Dimension records	Original	Hot key replicated
Implementation	Simple	More complex
Best use	Balanced data	Severe data skew

Production insight: Salting is not automatically faster. It adds dimension-table replication and additional processing. For smaller dimension tables, a broadcast join may be a better option. Also, Databricks Runtime supports Adaptive Query Execution (AQE), which can automatically mitigate certain skewed joins.
'''
```

```python
# Problem 27 : Optimize PySpark Joins using Broadcast Joins

'''

Concepts: Broadcast joins, shuffle joins, broadcast hints, physical execution plans, Spark optimization.
1. Business scenario
Imagine you're working on an e-commerce data pipeline.

You have two DataFrames:
- orders_df: 10 million order records.
- customers_df: 5,000 customer records.
You need to join them using customer_id.

A normal join may involve a shuffle, where Spark redistributes data across partitions.

But there's a better approach when one DataFrame is sufficiently small: Broadcast Join.
'''

# Create sample DataFrames
from pyspark.sql import functions as F
from pyspark.sql.functions import broadcast

orders_data_27 = [
    (1, 101, 5000),
    (2, 102, 3000),
    (3, 103, 7000),
    (4, 101, 2000),
    (5, 104, 4000),
    (6, 102, 6000)
]

orders_df_27 = spark.createDataFrame(
    orders_data_27,
    ["order_id", "customer_id", "amount"]
)

customers_data_27 = [
    (101, "Rahul"),
    (102, "Priya"),
    (103, "Aman"),
    (104, "Sneha")
]

customers_df_27 = spark.createDataFrame(
    customers_data_27,
    ["customer_id", "customer_name"]
)
```

```python
# Perform a normal join
normal_join_27 = orders_df_27.join(
    customers_df_27,
    on="customer_id",
    how="inner"
)

normal_join_27.show()

# This produces the correct result, but depending on table sizes, Spark may choose a shuffle-based join.

# Perform a broadcast join

broadcast_join_27 = orders_df_27.join(
    broadcast(customers_df_27),
    on="customer_id",
    how="inner"
)

broadcast_join_27.show()

'''
What happens internally?
1. Spark collects the small DataFrame for broadcasting.
2. It distributes the small DataFrame to the executors.
3. Each executor joins its local portion of the large DataFrame with the broadcast data.
4. The large DataFrame generally avoids being shuffled for this join.
This can significantly reduce network communication.
'''
```

```python
'''
The actual plan depends on Spark configuration, table statistics, and runtime optimization.
6. Important interview comparison
Feature	Broadcast Join	Sort-Merge Join
Best suited for	One small table	Large tables
Shuffle	Avoids shuffling the large side for the join	Usually shuffles both sides
Sorting	Not required for the join	Requires sorting
Memory	Small table must fit in executor memory	Handles larger datasets
Performance	Often faster for suitable small tables	Useful for large-large joins

Production considerations:
- Broadcasting a huge DataFrame can cause executor memory pressure or failures.
- Spark can automatically select broadcast joins based on statistics and spark.sql.autoBroadcastJoinThreshold.
- broadcast() is a hint; always inspect the physical plan to confirm the strategy.
- AQE can change the join strategy at runtime when enabled.

'''
```

```python
# Problem 28 — Adaptive Query Execution (AQE) in PySpark

'''
Objective: Understand how Spark dynamically optimizes query execution using runtime statistics.
1. Business scenario
Imagine you're processing millions of e-commerce transactions.

Your Spark job initially creates 200 shuffle partitions. However, after processing, Spark discovers:
- Some partitions contain very little data.
- Some partitions contain significantly more data.
- A join could execute more efficiently using a different strategy.
AQE allows Spark to optimize the physical execution plan using statistics collected during execution.

It has three major capabilities:
1. Coalescing shuffle partitions.
2. Optimizing skewed joins.
3. Dynamically changing join strategies.

'''

# Create DataFrame
from pyspark.sql import functions as F

transactions_data_28 = [
    (1, "Mumbai", 1000),
    (2, "Mumbai", 2000),
    (3, "Pune", 1500),
    (4, "Delhi", 3000),
    (5, "Mumbai", 2500),
    (6, "Pune", 1800),
    (7, "Delhi", 2200),
    (8, "Mumbai", 1200),
    (9, "Pune", 1700),
    (10, "Delhi", 2800)
]

transactions_df_28 = spark.createDataFrame(
    transactions_data_28,
    ["transaction_id", "city", "amount"]
)

transactions_df_28.show()
```

```python
# Create shuffle partitions
partitioned_df_28 = transactions_df_28.repartition(
    20,
    "city"
)

print("Repartitioning completed successfully")
# Why repartition?

# repartition() redistributes data across partitions. Here, we're deliberately creating 20 partitions to demonstrate shuffle behavior.

# Inspect partition distribution without RDD
# Use Spark's built-in spark_partition_id() function:

partition_distribution_28 = (
    partitioned_df_28
    .withColumn(
        "partition_id",
        F.spark_partition_id()
    )
    .groupBy("partition_id")
    .count()
    .orderBy("partition_id")
)

partition_distribution_28.show()

# This shows how many records are present in each partition.

# Why this works: spark_partition_id() is a DataFrame function, so it avoids accessing the unsupported RDD API.

# Perform aggregation

result_df_28 = partitioned_df_28.groupBy("city").agg(
    F.sum("amount").alias("total_amount"),
    F.count("*").alias("total_transactions")
)

result_df_28.show()
```

```python
# Problem 31 — Delta Lake Schema Evolution using PySpark
'''
Today's concept: How to handle changing source schemas without breaking your ETL pipeline.

We'll use only PySpark DataFrame APIs and operations supported by your Databricks Serverless environment. No DBFS paths or SQL cells.
1. Business scenario
Imagine you're maintaining a customer Delta table.

Initially, your source contains:
- customer_id
- name
- city
After a few months, the business introduces a new column: loyalty_tier.

Your existing Delta table doesn't have this column.

How do you add the new column without recreating the entire table?

This is called Schema Evolution.

'''

# Create the initial Delta table
from pyspark.sql import functions as F

customer_data_31 = [
    (101, "Rahul", "Mumbai"),
    (102, "Priya", "Pune"),
    (103, "Aman", "Delhi")
]

columns_31 = ["customer_id", "name", "city"]

customer_df_31 = spark.createDataFrame(
    customer_data_31,
    columns_31
)

customer_df_31.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("customer_schema_31")


initial_df_31 = spark.table("customer_schema_31")

initial_df_31.show()
initial_df_31.printSchema()
```

```python
# New source arrives with an additional column
new_customer_data_31 = [
    (104, "Sneha", "Mumbai", "Gold"),
    (105, "Rohan", "Bangalore", "Silver")
]

new_columns_31 = [
    "customer_id",
    "name",
    "city",
    "loyalty_tier"
]

new_customer_df_31 = spark.createDataFrame(
    new_customer_data_31,
    new_columns_31
)

new_customer_df_31.show()

# Notice that the incoming DataFrame has one additional column: loyalty_tier.
```

```python
# Enable schema evolution during append
new_customer_df_31.write \
    .format("delta") \
    .mode("append") \
    .option("mergeSchema", "true") \
    .saveAsTable("customer_schema_31")
# What does mergeSchema do?

# It tells Delta Lake to evolve the target table schema to accommodate compatible new columns.

# Read the updated Delta table
final_customer_df_31 = spark.table(
    "customer_schema_31"
)

final_customer_df_31.show()

final_customer_df_31.printSchema()

'''
Why are the first three records showing NULL?

Because those records existed before loyalty_tier was introduced.

Delta Lake preserves the existing records and returns NULL for the newly added column where no value existed.
6. Important interview concepts
Concept	Meaning
Schema evolution	Automatically accommodate compatible schema changes
mergeSchema	Merge incoming schema with target schema during supported writes
append	Add incoming records
overwrite	Replace existing table data
NULL for old records	No historical value was supplied

Production insight: Schema evolution should be controlled. Don't blindly accept every incoming column change. Validate data types, required fields and unexpected schema changes before writing.

'''

# mergeSchema changes the schema; it does not perform deduplication or upsert.
```

```python
# Deduplication using Window Functions in PySpark
'''

Today's objective: Remove duplicate customer records while retaining the latest updated record.

This is a very common Data Engineering interview problem.
1. Business scenario
Your customer source system sends multiple records for the same customer.

You need to retain only the latest record based on updated_at.

Example:
customer_id	name	city	updated_at
101	Rahul	Mumbai	2026-01-01
101	Rahul	Pune	2026-02-01
102	Priya	Delhi	2026-01-15
102	Priya	Mumbai	2026-02-15
103	Aman	Pune	2026-01-20

Expected: Keep the latest record for each customer.

'''

# Create a DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

customer_data_33 = [
    (101, "Rahul", "Mumbai", "2026-01-01"),
    (101, "Rahul", "Pune", "2026-02-01"),
    (102, "Priya", "Delhi", "2026-01-15"),
    (102, "Priya", "Mumbai", "2026-02-15"),
    (103, "Aman", "Pune", "2026-01-20")
]

columns_33 = [
    "customer_id", "name", "city", "updated_at"
]

customer_df_33 = spark.createDataFrame(
    customer_data_33,
    columns_33
).withColumn(
    "updated_at",
    F.to_timestamp("updated_at")
)

customer_df_33.show()
```

```python
# Apply row_number()
# First, define a window partitioned by customer ID and ordered by latest update.
window_33 = Window.partitionBy(
    "customer_id"
).orderBy(
    F.col("updated_at").desc()
)

# Now assign row numbers:
ranked_df_33 = customer_df_33.withColumn(
    "row_num",
    F.row_number().over(window_33)
)

ranked_df_33.show()

#Keep only the latest record
deduplicated_df_33 = ranked_df_33.filter(
    F.col("row_num") == 1
).drop("row_num")

deduplicated_df_33.show()


'''

Interview concepts
- Window.partitionBy() groups records for window calculations.
- orderBy(desc()) puts the latest record first.
- row_number() assigns sequential numbers within each customer.
- Filtering row_num == 1 retains the latest record.
Production tip: If two records have identical timestamps, add a deterministic tie-breaker such as ingestion_timestamp or event_id.

'''
```

```python
# Problem 34 Handling NULL Values and Data Quality in PySpark
'''
Objective: Build a data-cleaning pipeline that identifies missing values, applies business rules, and separates rejected records.

We'll use only PySpark DataFrame APIs compatible with your Databricks Serverless environment.
1. Business scenario
You receive customer records from multiple source systems. Some records have:
- Missing customer IDs.
- Missing names.
- NULL cities.
- Missing salary values.
- Invalid or negative salaries.
Your task is to clean the data without losing track of rejected records.

'''

# Create a dataframe
from pyspark.sql import functions as F

customer_data_34 = [
    (101, "Rahul", "Mumbai", 50000),
    (102, "Priya", None, 60000),
    (103, None, "Delhi", 45000),
    (104, "Aman", "Pune", None),
    (None, "Sneha", "Mumbai", 55000),
    (106, "Rohan", "Delhi", -1000),
    (107, "Neha", None, None),
    (108, "Karan", "Pune", 70000)
]

customer_df_34 = spark.createDataFrame(
    customer_data_34,
    ["customer_id", "name", "city", "salary"]
)

customer_df_34.show()
```

```python
# Identify missing values
null_count_df_34 = customer_df_34.select(
    *[
        F.count(
            F.when(F.col(c).isNull(), 1)
        ).alias(c)
        for c in customer_df_34.columns
    ]
)

null_count_df_34.show()

# This calculates the NULL count for every column.

# Apply business rules
# Let's define a valid customer record:
# - customer_id must not be NULL.
# - name must not be NULL.
# - city must not be NULL.
# - salary must be greater than zero.

valid_condition_34 = (
    F.col("customer_id").isNotNull() &
    F.col("name").isNotNull() &
    F.col("city").isNotNull() &
    (F.col("salary") > 0)
)

valid_df_34 = customer_df_34.filter(
    valid_condition_34
)

rejected_df_34 = customer_df_34.filter(
    ~valid_condition_34
)

# Add rejection reasons

rejected_with_reason_34 = rejected_df_34.withColumn(
    "rejection_reason",
    F.concat_ws(
        ", ",
        F.when(
            F.col("customer_id").isNull(),
            F.lit("Missing customer ID")
        ),
        F.when(
            F.col("name").isNull(),
            F.lit("Missing name")
        ),
        F.when(
            F.col("city").isNull(),
            F.lit("Missing city")
        ),
        F.when(
            F.col("salary").isNull(),
            F.lit("Missing salary")
        ),
        F.when(
            F.col("salary") <= 0,
            F.lit("Invalid salary")
        )
    )
)

# Display results
print("VALID RECORDS")
valid_df_34.show()

print("REJECTED RECORDS")
rejected_with_reason_34.show(truncate=False)
```

```python
# Problem 40 PySpark — Running Total Using Window Functions

# Interview scenario: Calculate each customer's cumulative order amount, ordered by transaction date.

# This is a frequently asked Data Engineer interview problem involving Window, partitionBy(), and rowsBetween().

#  Create a DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

orders_data_40 = [
    (1, 101, 500, "2026-01-01"),
    (2, 101, 300, "2026-01-02"),
    (3, 101, 700, "2026-01-03"),
    (4, 102, 400, "2026-01-01"),
    (5, 102, 600, "2026-01-02"),
    (6, 102, 200, "2026-01-03")
]

orders_df_40 = spark.createDataFrame(
    orders_data_40,
    ["order_id", "customer_id", "amount", "order_date"]
).withColumn(
    "order_date", F.to_date("order_date")
)

orders_df_40.show()
```

```python
# Problem 41: PySpark — Running Count and Cumulative Average
# Interview scenario: For each customer, calculate:
# 1. Cumulative number of orders.
# 2. Cumulative average order amount.

# Create Dataframe
from pyspark.sql import functions as F
from pyspark.sql.window import Window

orders_data_41 = [
    (1, 101, 500, "2026-01-01"),
    (2, 101, 300, "2026-01-02"),
    (3, 101, 700, "2026-01-03"),
    (4, 102, 400, "2026-01-01"),
    (5, 102, 600, "2026-01-02"),
    (6, 102, 200, "2026-01-03")
]

orders_df_41 = spark.createDataFrame(
    orders_data_41,
    ["order_id", "customer_id", "amount", "order_date"]
).withColumn(
    "order_date", F.to_date("order_date")
)

# Define the window
window_spec_41 = (
    Window
    .partitionBy("customer_id")
    .orderBy("order_date", "order_id")
    .rowsBetween(
        Window.unboundedPreceding,
        Window.currentRow
    )
)


# Calculate both metrics
result_41 = orders_df_41 \
    .withColumn(
        "running_order_count",
        F.count("order_id").over(window_spec_41)
    ) \
    .withColumn(
        "running_average",
        F.avg("amount").over(window_spec_41)
    )

result_41.orderBy(
    "customer_id", "order_date"
).show()


'''
Understand the difference
Function	What it calculates
F.count().over()	Running count
F.sum().over()	Running total
F.avg().over()	Running average

Interview insight: A running average is not the same as an average of daily averages. Here, each order contributes equally to the cumulative average.
'''
```

```python
# Problem 42: PySpark — lag() and lead() Window Functions
# Interview scenario: Calculate the previous order amount, next order amount, and difference between the current and previous order for each customer.

# This is a frequently asked Data Engineer interview question, especially in transaction and time-series analytics.

# Create DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

orders_data_42 = [
    (1, 101, 500, "2026-01-01"),
    (2, 101, 300, "2026-01-02"),
    (3, 101, 700, "2026-01-03"),
    (4, 101, 900, "2026-01-04"),
    (5, 102, 400, "2026-01-01"),
    (6, 102, 600, "2026-01-02"),
    (7, 102, 200, "2026-01-03")
]

orders_df_42 = spark.createDataFrame(
    orders_data_42,
    ["order_id", "customer_id", "amount", "order_date"]
).withColumn(
    "order_date", F.to_date("order_date")
)

orders_df_42.show()
```

```python
# Problem 44: PySpark — Sessionization Using Window Functions
# Interview scenario: A website records user activities. You need to identify separate user sessions.

# A new session starts when the time difference between two consecutive activities is greater than 30 minutes.

# This is a common Data Engineering problem involving lag(), timestamp differences, and cumulative sums.

# Create a DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

events_data_44 = [
    (1, 101, "2026-01-01 09:00:00"),
    (2, 101, "2026-01-01 09:10:00"),
    (3, 101, "2026-01-01 09:35:00"),
    (4, 101, "2026-01-01 10:20:00"),
    (5, 101, "2026-01-01 10:30:00"),
    (6, 102, "2026-01-01 11:00:00"),
    (7, 102, "2026-01-01 11:15:00"),
    (8, 102, "2026-01-01 12:00:00")
]

events_df_44 = spark.createDataFrame(
    events_data_44,
    ["event_id", "user_id", "event_time"]
).withColumn(
    "event_time",
    F.to_timestamp("event_time")
)

events_df_44.show(truncate=False)
```

```python
# Define the window
window_spec_44 = Window.partitionBy(
    "user_id"
).orderBy(
    "event_time", "event_id"
)

# Find the previous event time
# Use lag() to get the previous activity timestamp.

step_1_44 = events_df_44.withColumn(
    "previous_event_time",
    F.lag("event_time").over(window_spec_44)
)

# Calculate the time difference

step_2_44 = step_1_44.withColumn(
    "time_diff_minutes",
    (
        F.col("event_time").cast("long") -
        F.col("previous_event_time").cast("long")
    ) / 60
)

# Identify new sessions
# A new session starts when:
# - There is no previous event.
# - The time difference is greater than 30 minutes.

step_3_44 = step_2_44.withColumn(
    "new_session",
    F.when(
        F.col("previous_event_time").isNull() |
        (F.col("time_diff_minutes") > 30),
        1
    ).otherwise(0)
)

# Generate session IDs
# Use cumulative sum() to assign session numbers.
session_window_44 = window_spec_44.rowsBetween(
    Window.unboundedPreceding,
    Window.currentRow
)

result_44 = step_3_44.withColumn(
    "session_id",
    F.sum("new_session").over(session_window_44)
)

result_44.orderBy(
    "user_id", "event_time"
).show(truncate=False)
```

```python
# Problem 45: PySpark — Session-Level Aggregations

#Interview scenario: You have user activity events with session IDs. Calculate session-level metrics:

# 1. Total events per session.
# 2. Session start time.
# 3. Session end time.
# 4. Session duration in minutes.

#This is a practical Data Engineering problem involving groupBy(), count(), min(), max(), and timestamp calculations.


# Create DataFrame
from pyspark.sql import functions as F

events_data_45 = [
    (1, 101, 1, "2026-01-01 09:00:00"),
    (2, 101, 1, "2026-01-01 09:10:00"),
    (3, 101, 1, "2026-01-01 09:35:00"),
    (4, 101, 2, "2026-01-01 10:20:00"),
    (5, 101, 2, "2026-01-01 10:30:00"),
    (6, 102, 1, "2026-01-01 11:00:00"),
    (7, 102, 1, "2026-01-01 11:15:00"),
    (8, 102, 2, "2026-01-01 12:00:00")
]

events_df_45 = spark.createDataFrame(
    events_data_45,
    ["event_id", "user_id", "session_id", "event_time"]
).withColumn(
    "event_time", F.to_timestamp("event_time")
)

events_df_45.show(truncate=False)
```

```python
# Problem 46: PySpark — Date-wise Sales Analysis

#Interview scenario: You have an orders dataset. Calculate daily sales metrics and compare each day's revenue with the previous day's revenue.

# You need to calculate:
# 1. Total daily revenue.
# 2. Total daily orders.
# 3. Previous day's revenue.
# 4. Day-over-day revenue change.
# 5. Day-over-day percentage change.
# This is a common Data Engineer and Data Analyst interview problem.

# Create a DataFrame

from pyspark.sql import functions as F
from pyspark.sql.window import Window

sales_data_46 = [
    (1, 101, 500, "2026-01-01"),
    (2, 102, 300, "2026-01-01"),
    (3, 101, 700, "2026-01-02"),
    (4, 103, 200, "2026-01-02"),
    (5, 102, 400, "2026-01-03"),
    (6, 103, 600, "2026-01-03"),
    (7, 101, 900, "2026-01-04")
]

sales_df_46 = spark.createDataFrame(
    sales_data_46,
    ["order_id", "customer_id", "amount", "order_date"]
).withColumn(
    "order_date", F.to_date("order_date")
)

sales_df_46.show()
```

```python
# Problem 47: PySpark — Handling Missing Dates in Time-Series Data

# Interview scenario: You have daily sales data, but some dates are missing. Your task is to generate a continuous date range and fill missing sales with zero.

# Create a DataFrame

from pyspark.sql import functions as F

sales_data_47 = [
    ("2026-01-01", 1000),
    ("2026-01-02", 1500),
    ("2026-01-05", 2000),
    ("2026-01-07", 1200)
]

sales_df_47 = spark.createDataFrame(
    sales_data_47,
    ["sales_date", "revenue"]
).withColumn(
    "sales_date", F.to_date("sales_date")
)

sales_df_47.show()

# Notice that January 3, 4 and 6 are missing.

# Generate a continuous date range

#First, find the minimum and maximum dates.

date_range_47 = sales_df_47.agg(
    F.min("sales_date").alias("min_date"),
    F.max("sales_date").alias("max_date")
)

date_range_47.show()

# Generate all dates between them:

all_dates_47 = date_range_47.select(
    F.explode(
        F.sequence(
            F.col("min_date"),
            F.col("max_date"),
            F.expr("INTERVAL 1 DAY")
        )
    ).alias("sales_date")
)

all_dates_47.show()

# Left join with original sales

complete_sales_47 = all_dates_47.join(
    sales_df_47,
    on="sales_date",
    how="left"
)

# Replace missing revenue with zero
result_47 = complete_sales_47.withColumn(
    "revenue",
    F.coalesce(F.col("revenue"), F.lit(0))
).orderBy("sales_date")

result_47.show()

```

```python
# Problem 49: PySpark — SCD Type 2 (Historical Data Tracking)

# Interview scenario: A customer changes their city or account status over time. Your data pipeline must preserve the historical records instead of overwriting the old values.

# This is called Slowly Changing Dimension (SCD) Type 2 and is widely used in data warehouses.

# Create a DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

customers_data_49 = [
    (101, "Aman", "Mumbai", "Active", "2026-01-01"),
    (101, "Aman", "Mumbai", "Active", "2026-01-03"),
    (101, "Aman", "Pune", "Active", "2026-01-05"),
    (101, "Aman", "Pune", "Inactive", "2026-01-08"),
    (102, "Rahul", "Pune", "Active", "2026-01-02"),
    (102, "Rahul", "Mumbai", "Active", "2026-01-06")
]

customers_df_49 = spark.createDataFrame(
    customers_data_49,
    ["customer_id", "name", "city", "status", "updated_at"]
).withColumn(
    "updated_at", F.to_timestamp("updated_at")
)

customers_df_49.show()

# Identify changes using lag()
window_spec_49 = Window.partitionBy(
    "customer_id"
).orderBy("updated_at")

step_1_49 = customers_df_49 \
    .withColumn(
        "previous_city",
        F.lag("city").over(window_spec_49)
    ) \
    .withColumn(
        "previous_status",
        F.lag("status").over(window_spec_49)
    )
# Mark changed records
# A new version is created when the city or status changes.


step_2_49 = step_1_49.withColumn(
    "is_changed",
    F.when(
        F.col("previous_city").isNull() |
        (F.col("city") != F.col("previous_city")) |
        (F.col("status") != F.col("previous_status")),
        1
    ).otherwise(0)
)

# Retain only changed records

changed_records_49 = step_2_49.filter(
    F.col("is_changed") == 1
)
```

```python
# Generate version numbers
version_window_49 = Window.partitionBy(
    "customer_id"
).orderBy("updated_at")

versioned_df_49 = changed_records_49.withColumn(
    "version_no",
    F.row_number().over(version_window_49)
)

# Calculate effective dates
# Use lead() to identify when the next version begins. 

effective_window_49 = Window.partitionBy(
    "customer_id"
).orderBy("updated_at")

result_49 = versioned_df_49 \
    .withColumn(
        "effective_from",
        F.col("updated_at")
    ) \
    .withColumn(
        "effective_to",
        F.lead("updated_at").over(effective_window_49)
    ) \
    .withColumn(
        "is_current",
        F.when(F.col("effective_to").isNull(), True)
         .otherwise(False)
    ) \
    .select(
        "customer_id",
        "name",
        "city",
        "status",
        "version_no",
        "effective_from",
        "effective_to",
        "is_current"
    )

result_49.orderBy(
    "customer_id", "version_no"
).show(truncate=False)

'''
Understand SCD Type 1 vs Type 2
Feature	SCD Type 1	SCD Type 2
Historical data	Overwritten	Preserved
New records	Updated in place	New version created
Version tracking	No	Yes
Effective dates	Usually no	Yes
Use case	Correcting incorrect data	Tracking customer history

Your checkpoint
A customer's city changes from Mumbai to Pune.

'''
```

```python
# Problem 50: PySpark — End-to-End Incremental Data Pipeline
# Interview scenario: You receive daily customer transactions. Your task is to clean the data, remove duplicates, calculate customer-level metrics, and identify valid and rejected records.

# Create a DataFrame
from pyspark.sql import functions as F
from pyspark.sql.window import Window

transactions_data_50 = [
    (1, 101, 500, "COMPLETED", "2026-01-01"),
    (2, 101, 300, "COMPLETED", "2026-01-02"),
    (3, 102, 700, "COMPLETED", "2026-01-01"),
    (4, 102, 200, "CANCELLED", "2026-01-02"),
    (5, 101, 500, "COMPLETED", "2026-01-01"),
    (6, 103, 400, "COMPLETED", "2026-01-03"),
    (7, 104, -100, "COMPLETED", "2026-01-03"),
    (8, 105, 600, "PENDING", "2026-01-04")
]

transactions_df_50 = spark.createDataFrame(
    transactions_data_50,
    ["transaction_id", "customer_id", "amount", "status", "transaction_date"]
).withColumn(
    "transaction_date", F.to_date("transaction_date")
)

transactions_df_50.show()
```

```python
# Step 1 — Data quality validation

# Valid records must satisfy:
# - customer_id cannot be NULL.
# - amount must be greater than zero.
# - status must be COMPLETED or CANCELLED


validated_df_50 = transactions_df_50.withColumn(
    "is_valid",
    F.when(
        F.col("customer_id").isNotNull() &
        (F.col("amount") > 0) &
        F.col("status").isin("COMPLETED", "CANCELLED"),
        True
    ).otherwise(False)
)

valid_df_50 = validated_df_50.filter(
    F.col("is_valid") == True
)

rejected_df_50 = validated_df_50.filter(
    F.col("is_valid") == False
)

valid_df_50.show()
rejected_df_50.show()

# Step 2 — Remove duplicate transactions

# Keep only one record per transaction ID.


dedup_df_50 = valid_df_50.dropDuplicates(
    ["transaction_id"]
)

dedup_df_50.show()

# Here, the duplicate transaction ID 1 is reduced to one record.

# For production pipelines with multiple versions, prefer row_number() ordered by an update or ingestion timestamp to control which record survives.

# Calculate customer metrics
# Calculate total transactions, completed transactions, and completed revenue.

customer_metrics_50 = dedup_df_50.groupBy(
    "customer_id"
).agg(
    F.count("transaction_id").alias("total_transactions"),

    F.sum(
        F.when(F.col("status") == "COMPLETED", 1).otherwise(0)
    ).alias("completed_transactions"),

    F.sum(
        F.when(F.col("status") == "CANCELLED", 1).otherwise(0)
    ).alias("cancelled_transactions"),

    F.sum(
        F.when(F.col("status") == "COMPLETED", F.col("amount"))
         .otherwise(0)
    ).alias("total_revenue")
)

customer_metrics_50.orderBy("customer_id").show()
```

```python
# Calculate customer ranking
# Rank customers by their completed revenue.


ranking_window_50 = Window.orderBy(
    F.col("total_revenue").desc()
)

final_df_50 = customer_metrics_50.withColumn(
    "revenue_rank",
    F.dense_rank().over(ranking_window_50)
)

final_df_50.orderBy("revenue_rank").show()
```
